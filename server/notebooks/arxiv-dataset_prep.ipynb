# Preprocessing

## First we need to filter categories to get papers realted to neural network architectures,

```python
unique_categories = df["categories"].unique()
print(unique_categories)
```

```python
df["categories_list"] = df["categories"].str.split()
df
```

```python
NEURAL_NETWORK_CATEGORIES = {
  # type(df) TIER 1: Essential
  "cs.LG",  # Machine Learning - PRIMARY CATEGORY (90% of NN papers)
  "cs.CV",  # Computer Vision - CNNs, Vision Transformers
  "cs.CL",  # Computation and Language (NLP) - Transformers, BERT, GPT
  "cs.AI",  # Artificial Intelligence - General AI applications
  "cs.NE",  # Neural and Evolutionary Computing - Specifically NNs
  # TIER 2: Important - Should include for comprehensive coverage
  # 'stat.ML',  # Statistics ML - Statistical learning theory
  # 'cs.RO',    # Robotics - RL, neural control
  # 'cs.SD',    # Sound - Audio neural networks, speech
  # 'cs.IR',    # Information Retrieval - Neural ranking, recommendations
  # 'cs.SI',    # Social and Information Networks - Graph Neural Networks
  # 'cs.CR',    # Cryptography and Security - Adversarial ML
  # 'cs.HC',    # Human-Computer Interaction - Neural interfaces
  # 'cs.MM',    # Multimedia - Video/image processing with NNs
}


def filter_categories(cat_list):
  """
  Check if:
  1. Has at least one neural network category
  2. ALL categories start with cs., math., or stat.
  """
  if not cat_list:
    return False

  # Check 1: Must have at least one neural network category
  has_nn_category = any(cat in NEURAL_NETWORK_CATEGORIES for cat in cat_list)

  # Check 2 (Optional): ALL categories must start with cs. or stat.
  all_valid_prefixes = all(cat.startswith("cs.") or cat.startswith("stat.") for cat in cat_list)

  return has_nn_category and all_valid_prefixes
```

```python
mask = df["categories_list"].apply(filter_categories)
df = df[mask]
df
```

```python
from typing import List


def flatten_and_format_authors(nested_authors: list[list[str]]) -> list[str]:
  """
  Flattens a list of lists of author parts (e.g., [[Last, First], ...])
  into a single list of formatted full names (e.g., [First Last, ...]).
  """
  formatted_authors = []
  for author_parts in nested_authors:
    # Concatenate the elements of the sublist into a single string
    # using a space as a delimiter, and strip any leading/trailing spaces.
    full_name = " ".join(author_parts).strip()
    if full_name:
      formatted_authors.append(full_name)
  return formatted_authors


# Apply the function to the entire column to create the new 'authors_list' column
# we need to format authors list since it's nested structure isnt handled well by compression
merged_df["authors_list"] = merged_df["authors_parsed"].apply(flatten_and_format_authors)
merged_df = merged_df.drop(["authors_parsed"], axis=1)
merged_df
```

```python
import os
import random
import time

import pandas as pd
import requests
from tqdm import tqdm

# --- Config ---
OPENALEX_MAX_RPS = 10  # target requests per second
MAX_RETRIES = 4
CHECKPOINT_EVERY = 500
CHECKPOINT_FILE = "citations_checkpoint.parquet"


def fetch_citation_count(arxiv_id, email="you@email.com", delay=0.11):
  """Single-paper citation fetcher with adaptive back-off."""
  headers = {"User-Agent": "arxiv-citation-fetcher", "mailto": email}
  urls = [
    f"https://api.openalex.org/works/https://doi.org/10.48550/arXiv.{arxiv_id}",
    f"https://api.openalex.org/works/https://arxiv.org/abs/{arxiv_id}",
  ]

  for attempt in range(MAX_RETRIES):
    for url in urls:
      try:
        r = requests.get(url, headers=headers, timeout=8)
        if r.status_code == 200:
          j = r.json()
          return {
            "citations": j.get("cited_by_count", 0),
            "source": "openalex",
            "success": True,
          }
        if r.status_code == 404:
          continue
        if r.status_code == 429:  # hit the limit
          wait = int(r.headers.get("Retry-After", 2))
          print(f"[429] wait {wait}s for {arxiv_id}")
          time.sleep(wait)
          continue
        if r.status_code >= 500:
          time.sleep(1.5)
          continue
      except Exception as e:
        print(f"[error] {arxiv_id}: {e}")
        time.sleep(1)
    time.sleep(0.5)  # brief pause between retries
  # fallback
  try:
    s2 = f"https://api.semanticscholar.org/graph/v1/paper/ARXIV:{arxiv_id}"
    r = requests.get(s2, params={"fields": "citationCount"}, timeout=8)
    if r.status_code == 200:
      j = r.json()
      return {
        "citations": j.get("citationCount", 0),
        "source": "semantic_scholar",
        "success": True,
      }
  except Exception:
    pass
  return {"citations": 0, "source": None, "success": False}


def add_citations_fast(df, id_col="id", email="you@email.com"):
  """Sequential fetcher tuned for ~10 req/s with checkpointing."""
  if "citations" not in df:
    df["citations"] = 0
  if "citation_source" not in df:
    df["citation_source"] = ""

  start = 0
  if os.path.exists(CHECKPOINT_FILE):
    df = pd.read_parquet(CHECKPOINT_FILE)
    start = (df["citations"] > 0).sum()
    print(f"Resuming from index {start}")

  delay = 1 / OPENALEX_MAX_RPS
  pbar = tqdm(range(start, len(df)), desc="Citations (~10 req/s)")
  for i in pbar:
    arxiv_id = str(df.at[i, id_col])
    res = fetch_citation_count(arxiv_id, email=email, delay=delay)
    df.at[i, "citations"] = res["citations"]
    df.at[i, "citation_source"] = res["source"] or ""
    # base delay plus tiny jitter
    time.sleep(delay + random.uniform(0.01, 0.03))
    # save checkpoint
    if (i + 1) % CHECKPOINT_EVERY == 0:
      df.to_parquet(CHECKPOINT_FILE, index=False)
      pbar.set_postfix({"checkpoint": f"{i + 1}"})
  df.to_parquet("arxiv_with_citations.parquet", index=False)
  print("✅ Saved arxiv_with_citations.parquet")
  return df
```
